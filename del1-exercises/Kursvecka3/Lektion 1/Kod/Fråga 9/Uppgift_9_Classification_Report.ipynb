````python
# Imports
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix

print(" Imports klara!")
````

## Kör den ursprungliga koden

````python
# Ursprunglig kod från uppgiften
y_true = [0, 1, 2, 2, 2]
y_pred = [0, 0, 2, 2, 1]
target_names = ['class 0', 'class 1', 'class 2']

print("Classification Report:")
print(classification_report(y_true, y_pred, target_names=target_names))
````

## Visualisera konfusionsmatris

````python
# Skapa konfusionsmatris för visualisering
cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=target_names, yticklabels=target_names)
plt.title('Konfusionsmatris', fontsize=14, fontweight='bold')
plt.ylabel('Faktisk klass', fontsize=12)
plt.xlabel('Predikterad klass', fontsize=12)
plt.show()

print("Konfusionsmatris:")
print(cm)
````

## Förklaring av resultatet

````python
print("=== Förklaring av Classification Report ===\n")

print("Class 0:")
print("- Faktiskt: 1 gång")
print("- Predikterat: 2 gånger (1 rätt, 1 fel)")
print("- Precision: 1/2 = 0.50")
print("- Recall: 1/1 = 1.00")

print("\nClass 1:")
print("- Faktiskt: 1 gång")
print("- Predikterat: 1 gång (0 rätt, 1 fel)")
print("- Precision: 0/1 = 0.00")
print("- Recall: 0/1 = 0.00")

print("\nClass 2:")
print("- Faktiskt: 3 gånger")
print("- Predikterat: 2 gånger (2 rätt, 0 fel)")
print("- Precision: 2/2 = 1.00")
print("- Recall: 2/3 = 0.67")

print("\n=== Sammanfattning ===")
print("• Precision: Hur många av de predikterade som var rätt")
print("• Recall: Hur många av de faktiska som hittades")
print("• F1-score: Harmoniska medelvärdet av precision och recall")
print("• Support: Antal samples per klass")
````
