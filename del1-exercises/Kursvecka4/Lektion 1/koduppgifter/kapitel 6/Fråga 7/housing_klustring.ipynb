# Kapitel 6 - Koduppgift 7: Housing Klustring

I denna uppgift använder vi datasetet 'housing.csv' för att genomföra klustring på median_income, latitude och longitude.

## Importera bibliotek och ladda data

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
import numpy as np

# Sätt stil för plots
plt.style.use('default')
sns.set_palette('husl')

print('Bibliotek importerade!')
```

## Ladda och utforska data

```python
# Ladda housing dataset
df = pd.read_csv('housing.csv')

print(f'Dataset shape: {df.shape}')
print(f'\nKolumner: {list(df.columns)}')
print(f'\nFörsta 5 rader:')
print(df.head())
```

```python
# Grundläggande information
print('Dataset info:')
print(df.info())
print(f'\nSaknade värden:')
print(df.isnull().sum())
print(f'\nBeskrivande statistik:')
print(df.describe())
```

## Förbered data för klustring

```python
# Välj features för klustring
X = df.loc[:, ['median_income', 'latitude', 'longitude']]

print(f'Features shape: {X.shape}')
print(f'\nFörsta 5 rader:')
print(X.head())

print(f'\nBeskrivande statistik för features:')
print(X.describe())
```

## Standardisera data

```python
# Standardisera features (viktigt för klustring)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(f'Data efter standardisering: {X_scaled.shape}')
print(f'\nGenomsnitt: {X_scaled.mean(axis=0)}')
print(f'Standardavvikelse: {X_scaled.std(axis=0)}')
```

## Välja optimalt antal kluster

```python
# Testa olika antal kluster
k_range = range(2, 11)
inertias = []
silhouette_scores = []

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    
    inertias.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(X_scaled, kmeans.labels_))

print('Inertia och Silhouette scores:')
for k, inertia, sil in zip(k_range, inertias, silhouette_scores):
    print(f'k={k}: Inertia={inertia:.2f}, Silhouette={sil:.3f}')
```

```python
# Visualisera Elbow method och Silhouette scores
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Elbow method
axes[0].plot(k_range, inertias, 'bo-')
axes[0].set_title('Elbow Method')
axes[0].set_xlabel('Antal kluster (k)')
axes[0].set_ylabel('Inertia')
axes[0].grid(True, alpha=0.3)

# Silhouette scores
axes[1].plot(k_range, silhouette_scores, 'ro-')
axes[1].set_title('Silhouette Score')
axes[1].set_xlabel('Antal kluster (k)')
axes[1].set_ylabel('Silhouette Score')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Klustring med 6 kluster (som i originalkoden)

```python
# K-means med 6 kluster
kmeans = KMeans(n_clusters=6, random_state=42, n_init=10)
X['Cluster'] = kmeans.fit_predict(X_scaled)
X['Cluster'] = X['Cluster'].astype('category')

print(f'Klustring slutförd!')
print(f'Antal kluster: {len(X["Cluster"].unique())}')
print(f'\nKlusterfördelning:')
print(X['Cluster'].value_counts().sort_index())
print(f'\nFörsta 5 rader:')
print(X.head())
```

## Visualisera klustring

```python
# Visualisera klustring på karta
plt.figure(figsize=(12, 8))
sns.relplot(
    x='longitude', y='latitude', hue='Cluster', data=X, 
    height=8, aspect=1.2, palette='husl'
)
plt.title('Housing Klustring - Geografisk fördelning')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.grid(True, alpha=0.3)
plt.show()
```

## Analysera kluster

```python
# Analysera varje kluster
print('Klusteranalys:')
print('=' * 50)

for cluster in sorted(X['Cluster'].unique()):
    cluster_data = X[X['Cluster'] == cluster]
    
    print(f'\nKluster {cluster}:')
    print(f'  Antal områden: {len(cluster_data)}')
    print(f'  Genomsnittlig median_income: {cluster_data["median_income"].mean():.2f}')
    print(f'  Latitude range: {cluster_data["latitude"].min():.2f} - {cluster_data["latitude"].max():.2f}')
    print(f'  Longitude range: {cluster_data["longitude"].min():.2f} - {cluster_data["longitude"].max():.2f}')
```

## Experimentera med olika antal kluster

```python
# Testa olika antal kluster och jämför
best_k = k_range[np.argmax(silhouette_scores)]
print(f'Bästa k baserat på Silhouette score: {best_k}')

# Klustring med optimalt antal kluster
kmeans_optimal = KMeans(n_clusters=best_k, random_state=42, n_init=10)
X_optimal = X.copy()
X_optimal['Cluster_Optimal'] = kmeans_optimal.fit_predict(X_scaled)
X_optimal['Cluster_Optimal'] = X_optimal['Cluster_Optimal'].astype('category')

print(f'\nOptimal klustring med {best_k} kluster:')
print(X_optimal['Cluster_Optimal'].value_counts().sort_index())
```

```python
# Jämför 6 kluster vs optimalt antal
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# 6 kluster
sns.scatterplot(data=X, x='longitude', y='latitude', hue='Cluster', 
                palette='husl', ax=axes[0])
axes[0].set_title(f'6 Kluster (Original)')
axes[0].grid(True, alpha=0.3)

# Optimalt antal kluster
sns.scatterplot(data=X_optimal, x='longitude', y='latitude', 
                hue='Cluster_Optimal', palette='husl', ax=axes[1])
axes[1].set_title(f'{best_k} Kluster (Optimalt)')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
